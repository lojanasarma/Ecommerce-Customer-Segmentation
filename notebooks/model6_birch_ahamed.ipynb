This notebook trains the Birch model.

```
import pandas as pd
import os
from sklearn.cluster import Birch

os.makedirs("outputs/predictions", exist_ok=True)

df_rfm = pd.read_csv("outputs/2_rfm_data_herath.csv")
X = df_rfm[["Scaled_R", "Scaled_F", "Scaled_M"]]
```

```
print("Training Birch...")
model = Birch(n_clusters=4)
df_rfm["Cluster"] = model.fit_predict(X)
```

```
df_rfm[["CustomerID", "Cluster"]].to_csv("outputs/predictions/model6_birch_predictions.csv", index=False)
print("Predictions saved!")
```
