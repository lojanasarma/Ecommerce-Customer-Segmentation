This notebook trains the MiniBatch model.

```
import pandas as pd
import os
from sklearn.cluster import MiniBatchKMeans

os.makedirs("outputs/predictions", exist_ok=True)

df_rfm = pd.read_csv("outputs/2_rfm_data_herath.csv")
X = df_rfm[["Scaled_R", "Scaled_F", "Scaled_M"]]
```

```
print("Training MiniBatch KMeans...")
model = MiniBatchKMeans(n_clusters=4, random_state=42, batch_size=1024)
df_rfm["Cluster_MiniBatch"] = model.fit_predict(X)
```

```
df_rfm[["CustomerID", "Cluster_MiniBatch"]].to_csv("outputs/predictions/model2_minibatch_predictions.csv", index=False)
print("Predictions saved!")
```
