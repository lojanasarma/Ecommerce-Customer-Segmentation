This notebook trains the DBSCAN model.

```
import pandas as pd
import os
from sklearn.cluster import DBSCAN

os.makedirs("outputs/predictions", exist_ok=True)

df_rfm = pd.read_csv("outputs/2_rfm_data_herath.csv")
X = df_rfm[["Scaled_R", "Scaled_F", "Scaled_M"]]
```

```
print("Training DBSCAN...")
model = DBSCAN(eps=0.5, min_samples=10)
df_rfm["Cluster"] = model.fit_predict(X)
```

```
df_rfm[["CustomerID", "Cluster"]].to_csv("outputs/predictions/model5_dbscan_predictions.csv", index=False)
print("Predictions saved!")
```
